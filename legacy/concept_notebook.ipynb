Basic Monte Carlo to explore future outcome ranges.

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from simulation import monte_carlo_outcomes

df = monte_carlo_outcomes(trials=5000, mean=0.07, std=0.2, periods=12)
df.describe(percentiles=[0.1,0.5,0.9])
```
